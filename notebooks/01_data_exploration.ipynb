```python
import pandas as pd

df = pd.read_csv("../data/raw/UNSW_NB15_training-set.csv")

print(df.shape)
print(df.info())
print(df.isnull().sum())
```

```python
df.describe().T
```

```python
features = [
    "dur",
    "spkts",
    "dpkts",
    "sbytes",
    "dbytes",
    "rate",
    "sload",
    "dload",
    "smean",
    "dmean"
]

X = df[features].copy()



```

```python
import numpy as np

X_log = np.log1p(X)

```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X_log)
```

```python
processed_df = pd.DataFrame(
    X_scaled,
    columns=features
)

processed_df["id"] = df["id"].values
processed_df["attack_cat"] = df["attack_cat"].values
processed_df["label"] = df["label"].values

print(processed_df.shape)
print(processed_df.head())
```

```python
processed_df.to_csv(
    "../data/processed/processed_data.csv",
    index=False
)
```
